```python
# ==============================================================================
# Elevate Labs - AI & ML Internship
# Task 2: Exploratory Data Analysis (EDA)
# ==============================================================================

# 1. Import Essential Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual styles for clean, readable plots
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

print("--- Step 1: Libraries Imported Successfully ---")

# ==============================================================================
# 2. Load the Dataset
# ==============================================================================
# Automatically fetching the public Titanic dataset from GitHub
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("\n--- Step 2: Dataset Loaded ---")
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")

# Display the first 5 rows
print("First 5 rows of the dataset:")
display(df.head())

# ==============================================================================
# 3. Generate Summary Statistics & Data Info
# ==============================================================================
print("\n--- Step 3: Data Information & Summary Statistics ---")
print("\n--- Data Schema Info ---")
df.info()

print("\n--- Summary Statistics (Numeric Features) ---")
display(df.describe())

print("\n--- Summary Statistics (Categorical Features) ---")
display(df.describe(include=['O']))

print("\n--- Missing Values Count ---")
print(df.isnull().sum()[df.isnull().sum() > 0])

# ==============================================================================
# 4. Visualizations: Histograms & Boxplots (Numeric Features)
# ==============================================================================
print("\n--- Step 4: Generating Univariate Visualizations ---")

# Plotting the Distribution of Age (Histogram)
plt.figure(figsize=(10, 4))
sns.histplot(data=df, x='Age', kde=True, bins=30, color='skyblue')
plt.title('Distribution of Passenger Age (Detecting Skewness)')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

# Plotting Boxplot of Fare (Detecting Outliers)
plt.figure(figsize=(10, 4))
sns.boxplot(data=df, x='Fare', color='lightgreen')
plt.title('Boxplot of Passenger Fare (Detecting Outliers)')
plt.xlabel('Fare')
plt.show()

# ==============================================================================
# 5. Feature Relationships (Bivariate & Multivariate)
# ==============================================================================
print("\n--- Step 5: Generating Feature Relationship Visualizations ---")

# Barplot: Survival Rate by Passenger Class
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex', palette='muted')
plt.title('Survival Rate by Passenger Class and Gender')
plt.ylabel('Survival Probability')
plt.show()

# Boxplot: Distribution of Age across Ticket Classes
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Pclass', y='Age', palette='pastel')
plt.title('Age Distribution Across Passenger Classes')
plt.show()

# ==============================================================================
# 6. Correlation Matrix (Detecting Multicollinearity)
# ==============================================================================
print("\n--- Step 6: Correlation Matrix ---")

# Filter out non-numeric columns for correlation calculation
numeric_df = df.select_dtypes(include=[np.number])

plt.figure(figsize=(8, 6))
corr_matrix = numeric_df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt.show()

print("\n--- EDA Process Completed Successfully! ---")
```
